```python
epsilon = np.linspace(0.00001,8*np.pi,2000)
def eps_func(epsilon, epsilon_0):
    temp = np.sqrt(np.square(epsilon_0/epsilon) - 1)
    mask = temp<0
    temp[mask] = 0
    return temp
```

```python
x = np.arange(0, 8*np.pi, np.pi/2)
x_labels = [ f"{i:.4}" for i in x]

fix, ax = plt.subplots()
ax.set_xticks(x, x_labels, rotation='vertical')
ax.set_ylim(0,500)
#ax.set_xlim(0,4*np.pi)
plt.plot(epsilon, np.tan(epsilon))
plt.plot(epsilon, eps_func(epsilon, 1000))
```

```python
def func_cos(epsilon, epsilon_0):
    return np.tan(epsilon) - np.sqrt(np.square(epsilon_0/epsilon) - 1)
```

```python
cos_10 = lambda epsilon: func_cos(epsilon, 50)
```

```python
sol = root(cos_10, 4.5)
print(sol)
```
