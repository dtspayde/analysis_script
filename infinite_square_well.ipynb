```python
def psi(x, n=1):
    return np.sqrt(2)*np.sin(n*np.pi*x)

def prob_density(x, n=1):
    return psi(x,n)**2

prob_density(0.5) 
```

```python
x = np.linspace(0,1)
```

```python
plt.plot(x, psi(x), label="n=1")
#plt.plot(x, psi(x, 2), label="n=2")
#plt.plot(x, psi(x, 3), label="n=3")
plt.legend()
```

```python
plt.plot(x, prob_density(x), label="n=1")
#plt.plot(x, prob_density(x, 2), label="n=2")
#plt.plot(x, prob_density(x, 3), label="n=3")
```
